### Importación de datos



````python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
````

#1. Análisis de facturación



````python
ingreso_tienda1 = tienda["Precio"].sum()
ingreso_tienda2 = tienda2["Precio"].sum()
ingreso_tienda3 = tienda3["Precio"].sum()
ingreso_tienda4 = tienda4["Precio"].sum()

print("Ingreso Tienda 1:", ingreso_tienda1)
print("Ingreso Tienda 2:", ingreso_tienda2)
print("Ingreso Tienda 3:", ingreso_tienda3)
print("Ingreso Tienda 4:", ingreso_tienda4)
````

````python
import matplotlib.pyplot as plt

ingresos = [ingreso_tienda1, ingreso_tienda2, ingreso_tienda3, ingreso_tienda4]
tiendas = ["Tienda 1","Tienda 2","Tienda 3","Tienda 4"]

plt.bar(tiendas, ingresos)

plt.title("Ingresos por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")

plt.show()
````

# 2. Ventas por categoría

````python
categoria1 = tienda["Categoría del Producto"].value_counts()
categoria2 = tienda2["Categoría del Producto"].value_counts()
categoria3 = tienda3["Categoría del Producto"].value_counts()
categoria4 = tienda4["Categoría del Producto"].value_counts()

print("Tienda 1")
print(categoria1)

print("\nTienda 2")
print(categoria2)

print("\nTienda 3")
print(categoria3)

print("\nTienda 4")
print(categoria4)
````

````python
plt.pie(ingresos, labels=tiendas, autopct='%1.1f%%')

plt.title("Distribución de ingresos por tienda")

plt.show()
````

# 3. Calificación promedio de la tienda


````python
calificacion1 = tienda["Calificación"].mean()
calificacion2 = tienda2["Calificación"].mean()
calificacion3 = tienda3["Calificación"].mean()
calificacion4 = tienda4["Calificación"].mean()

print("Calificación Tienda 1:", calificacion1)
print("Calificación Tienda 2:", calificacion2)
print("Calificación Tienda 3:", calificacion3)
print("Calificación Tienda 4:", calificacion4)
````

````python
plt.plot(tiendas, ingresos, marker='o')

plt.title("Comparación de ingresos")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")

plt.show()
````

# 4. Productos más y menos vendidos

````python
productos1 = tienda["Producto"].value_counts()
productos2 = tienda2["Producto"].value_counts()
productos3 = tienda3["Producto"].value_counts()
productos4 = tienda4["Producto"].value_counts()

print("Más vendidos tienda 1")
print(productos1.head())

print("\nMenos vendidos tienda 1")
print(productos1.tail())
````

# 5. Envío promedio por tienda

````python
envio1 = tienda["Costo de envío"].mean()
envio2 = tienda2["Costo de envío"].mean()
envio3 = tienda3["Costo de envío"].mean()
envio4 = tienda4["Costo de envío"].mean()

print("Costo envío tienda 1:", envio1)
print("Costo envío tienda 2:", envio2)
print("Costo envío tienda 3:", envio3)
print("Costo envío tienda 4:", envio4)
````

# EXTRA : Análisis geográfico de las ventas

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(tienda["lon"], tienda["lat"], alpha=0.4, label="Tienda 1")
plt.scatter(tienda2["lon"], tienda2["lat"], alpha=0.4, label="Tienda 2")
plt.scatter(tienda3["lon"], tienda3["lat"], alpha=0.4, label="Tienda 3")
plt.scatter(tienda4["lon"], tienda4["lat"], alpha=0.4, label="Tienda 4")

plt.title("Distribución geográfica de ventas")
plt.xlabel("Longitud")
plt.ylabel("Latitud")

plt.legend()

plt.show()
````

## Análisis geográfico de las ventas (Extra)

En este análisis extra se usaron las coordenadas de latitud y longitud que vienen en los datos para ver en qué lugares se realizaron las compras. Con estas coordenadas se hizo un gráfico de dispersión que muestra los puntos donde se registraron las ventas de cada tienda.

Al observar el gráfico se puede notar que las ventas se concentran más en algunas zonas que en otras. Esto puede indicar que hay regiones donde los productos tienen más demanda o donde hay más clientes comprando.

Este tipo de análisis ayuda a entender mejor cómo se distribuyen las ventas según la ubicación geográfica y también podría servir para tomar decisiones en el futuro, como mejorar la distribución de productos o enfocarse en las zonas donde hay más ventas.
